```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create the dataset as a pandas DataFrame
data = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Weather': ['Sunny', 'Cloudy', 'Rainy', 'Sunny', 'Rainy', 'Cloudy', 'Sunny'],
    'Temperature': [32, 28, 25, 33, 26, 27, 34],  # in Celsius
    'Mood': [8, 6, 4, 9, 5, 6, 9]  # Mood scale 1-10
}

df = pd.DataFrame(data)
```

```python
print("Dataset:")
print(df)
```

```python
print("\nColumns in the dataset:", df.columns.tolist())
```

```python
print("Dimensions (rows, columns):", df.shape)
```

```python
print("\nInformation about dataset:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
df['Mood_Reverse'] = df['Mood'][::-1].values
print("\nDataset with Mood_Reverse column:")
print(df)
```

```python
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\nLast 5 rows:")
print(df.tail())
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['Day'], df['Mood'], marker='o', color='blue', linestyle='-', linewidth=2)
plt.title('College Mood Over Days')
plt.xlabel('Days')
plt.ylabel('Mood (1-10)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
colors = ['orange' if w=='Sunny' else 'gray' if w=='Cloudy' else 'blue' for w in df['Weather']]
plt.bar(df['Day'], df['Mood'], color=colors)
plt.title('College Mood vs Weather')
plt.xlabel('Days')
plt.ylabel('Mood (1-10)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Temperature'], df['Mood'], color='green', s=100)
for i, day in enumerate(df['Day']):
    plt.text(df['Temperature'][i]+0.2, df['Mood'][i]+0.1, day)
plt.title('Mood vs Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Mood (1-10)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Mood'], bins=[1,3,5,7,9,11], color='purple', edgecolor='black')
plt.title('Mood Frequency Distribution')
plt.xlabel('Mood (1-10)')
plt.ylabel('Frequency')
plt.show()
```

```python
weather_counts = df['Weather'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(weather_counts, labels=weather_counts.index, autopct='%1.1f%%',
        colors=['orange', 'gray', 'blue'], startangle=90)
plt.title('Weather Distribution in a Week')
plt.show()
```

```python
print("\nInsights from College Mood vs Weather Data:")
print("- College mood is higher on Sunny days.")
print("- Mood drops on Rainy days.")
print("- Mood is moderately better when temperature is higher.")
print("- Most common mood levels are 6, 8, and 9.")
print("- Sunny days are the majority in the week, followed by Cloudy and Rainy.")
```
